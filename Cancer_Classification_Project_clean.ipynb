```python
X = df.drop(columns=["target"])


y = df["target"].copy()
print("INPUT FEATURES (X)")
print("==================")
print("Shape:", X.shape)

print("\nTARGET VARIABLE (y)")
print("===================")
print("Shape:", y.shape)

print("\nTotal Input Features:", X.shape[1])
print("Total Target Records:", len(y))
```

```python

unique_labels = np.sort(y.unique())

print("Unique Target Labels:", unique_labels)

print("\nTarget Label Mapping:")

for label in unique_labels:
    print(
        f"{label} = "
        f"{cancer_data.target_names[int(label)].capitalize()}"
    )

assert set(unique_labels) == {0, 1}

print("\nTarget labels verified successfully!")
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("DATASET SPLIT COMPLETED SUCCESSFULLY!")
print("=====================================")
print("\nTraining Features:", X_train.shape)
print("Testing Features:", X_test.shape)

print("\nTraining Target:", y_train.shape)
print("Testing Target:", y_test.shape)

print("\nTotal Training Samples:", len(X_train))
print("Total Testing Samples:", len(X_test))
```

```python
class_distribution = pd.DataFrame({
    "Training Count": y_train.value_counts().reindex([0, 1]),
    "Testing Count": y_test.value_counts().reindex([0, 1]),
    "Training Percentage":
        y_train.value_counts(normalize=True).reindex([0, 1]) * 100,
    "Testing Percentage":
        y_test.value_counts(normalize=True).reindex([0, 1]) * 100
})

# Rename target labels
class_distribution.index = [
    cancer_data.target_names[i].capitalize()
    for i in class_distribution.index
]

print("CLASS DISTRIBUTION AFTER TRAIN-TEST SPLIT")
print("=========================================")

display(class_distribution.round(2))
```

```python
scaler = StandardScaler()


X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

print("FEATURE SCALING COMPLETED SUCCESSFULLY!")
print("=======================================")

print("\nScaled Training Data Shape:",
      X_train_scaled.shape)

print("Scaled Testing Data Shape:",
      X_test_scaled.shape)
```

```python

scaled_means = np.mean(X_train_scaled, axis=0)

scaled_stds = np.std(X_train_scaled, axis=0)

print("STANDARDIZATION VERIFICATION")
print("============================")

print("\nMeans of First 5 Features:")
print(np.round(scaled_means[:5], 4))

print("\nStandard Deviations of First 5 Features:")
print(np.round(scaled_stds[:5], 4))
```

```python

X_train_scaled_df = pd.DataFrame(
    X_train_scaled,
    columns=X.columns,
    index=X_train.index
)

print("FIRST FIVE STANDARDIZED TRAINING RECORDS")

display(X_train_scaled_df.head())
```

```python

print("FINAL PREPROCESSING VALIDATION")
print("==============================")


assert X_train_scaled.shape == (455, 30)
assert X_test_scaled.shape == (114, 30)

assert len(y_train) == 455
assert len(y_test) == 114

assert np.isfinite(X_train_scaled).all()
assert np.isfinite(X_test_scaled).all()

assert set(y_train.unique()) == {0, 1}
assert set(y_test.unique()) == {0, 1}

print("Training Data:", X_train_scaled.shape)
print("Testing Data:", X_test_scaled.shape)

print("\nAll validation checks passed!")
print("Data is ready for Neural Network development.")
```


# STEP 6: NEURAL NETWORK DEVELOPMENT

## 6.1 Introduction

In this stage, an Artificial Neural Network (ANN)
is developed using TensorFlow and Keras.

The model will learn patterns from the preprocessed
Breast Cancer Wisconsin dataset.

## 6.2 Neural Network Architecture

The model consists of:

- Input Layer: 30 Features
- Hidden Layer: 20 Neurons
- Hidden Activation: ReLU
- Output Layer: 2 Neurons
- Output Activation: Softmax

## 6.3 Model Configuration

- Optimizer: Adam
- Loss Function: Sparse Categorical Crossentropy
- Evaluation Metric: Accuracy

The developed model will be trained in the next stage.


```python
model = keras.Sequential([

    layers.Input(shape=(X_train_scaled.shape[1],)),


    layers.Dense(20, activation="relu", name="hidden_layer"),


    layers.Dense(2, activation="softmax", name="output_layer")

], name="Cancer_Classification_ANN")

print("Artificial Neural Network Created Successfully!")
```

```python

print("NEURAL NETWORK ARCHITECTURE")
print("===========================")

model.summary()
```

```python


model.compile(

    optimizer=keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="sparse_categorical_crossentropy",

    metrics=["accuracy"]

)

print("Neural Network Compiled Successfully!")
```

```python
print("MODEL CONFIGURATION VERIFICATION")
print("================================")

print("Model Name:", model.name)

print("Input Shape:", model.input_shape)

print("Output Shape:", model.output_shape)

print("Total Parameters:", model.count_params())

print("\nOptimizer:", type(model.optimizer).__name__)

print("Loss Function:", model.loss)

print("\nModel Verification Completed!")
```

```python
sample_input = X_train_scaled[:1]


sample_output = model(
    sample_input,
    training=False
).numpy()

print("MODEL INPUT SHAPE:", sample_input.shape)

print("\nMODEL OUTPUT PROBABILITIES:")

print("Malignant (0):", sample_output[0][0])

print("Benign (1):", sample_output[0][1])

print("\nSum of Probabilities:",
      sample_output.sum())

print("\nNeural Network Forward Pass Successful!")
```


# STEP 7: NEURAL NETWORK MODEL TRAINING

## 7.1 Introduction

In this stage, the Artificial Neural Network
developed in Step 6 will be trained using the
preprocessed Breast Cancer Wisconsin dataset.

The model will learn patterns from the numerical
tumor measurements to classify samples as
Malignant or Benign.

## 7.2 Training Configuration

- Optimizer: Adam
- Learning Rate: 0.001
- Epochs: 10
- Batch Size: 32
- Loss Function: Sparse Categorical Crossentropy
- Evaluation Metric: Accuracy

## 7.3 Training and Validation

The original training data will be divided into
training and validation subsets.

The reserved testing dataset will remain
untouched until the model evaluation stage.


```python

X_fit, X_val, y_fit, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.10,
    random_state=42,
    stratify=y_train
)

print("TRAINING AND VALIDATION SPLIT")
print("=============================")

print("Training Features:", X_fit.shape)
print("Validation Features:", X_val.shape)

print("Training Labels:", y_fit.shape)
print("Validation Labels:", y_val.shape)

print("\nTraining Samples:", len(X_fit))
print("Validation Samples:", len(X_val))
```

```python
training_scaler = StandardScaler()

X_fit_scaled = training_scaler.fit_transform(X_fit)

X_val_scaled = training_scaler.transform(X_val)

X_test_model_scaled = training_scaler.transform(X_test)

print("FINAL FEATURE SCALING COMPLETED")
print("================================")

print("Training Data:", X_fit_scaled.shape)
print("Validation Data:", X_val_scaled.shape)
print("Testing Data:", X_test_model_scaled.shape)

print("\nScaler fitted using training data only.")
```

```python
distribution = pd.DataFrame({
    "Training": y_fit.value_counts().reindex([0, 1]),
    "Validation": y_val.value_counts().reindex([0, 1])
})

distribution.index = [
    "Malignant",
    "Benign"
]

print("TRAINING AND VALIDATION CLASS DISTRIBUTION")
display(distribution)

print("\nTraining Percentages:")
print((y_fit.value_counts(normalize=True).sort_index() * 100).round(2))

print("\nValidation Percentages:")
print((y_val.value_counts(normalize=True).sort_index() * 100).round(2))
```

```python
print("STARTING NEURAL NETWORK TRAINING")
print("=================================")

history = model.fit(
    X_fit_scaled,
    y_fit.to_numpy(),
    validation_data=(
        X_val_scaled,
        y_val.to_numpy()
    ),
    epochs=10,
    batch_size=32,
    shuffle=True,
    verbose=1
)

print("\nNEURAL NETWORK TRAINING COMPLETED!")
```

```python
history_df = pd.DataFrame(history.history)

history_df.index = range(1, len(history_df) + 1)

history_df.index.name = "Epoch"

print("NEURAL NETWORK TRAINING HISTORY")
print("================================")

display(history_df.round(4))
```

```python
plt.figure(figsize=(9, 5))

plt.plot(
    history.history["accuracy"],
    marker="o",
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    marker="o",
    label="Validation Accuracy"
)

plt.title("Neural Network Training Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")

plt.legend()
plt.grid(True)
plt.xticks(range(10), range(1, 11))

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.plot(
    history.history["loss"],
    marker="o",
    label="Training Loss"
)
plt.plot(
    history.history["val_loss"],
    marker="o",
    label="Validation Loss"
)

plt.title("Neural Network Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.legend()
plt.grid(True)
plt.xticks(range(10), range(1, 11))

plt.tight_layout()
plt.show()
```

```python
final_train_accuracy = history.history["accuracy"][-1]
final_val_accuracy = history.history["val_accuracy"][-1]

final_train_loss = history.history["loss"][-1]
final_val_loss = history.history["val_loss"][-1]

print("FINAL NEURAL NETWORK TRAINING RESULTS")
print("=====================================")

print(f"Training Accuracy: {final_train_accuracy * 100:.2f}%")
print(f"Validation Accuracy: {final_val_accuracy * 100:.2f}%")

print(f"\nTraining Loss: {final_train_loss:.4f}")
print(f"Validation Loss: {final_val_loss:.4f}")

print("\nTotal Epochs:", len(history.history["accuracy"]))

print("\nModel Training Analysis Completed!")
```

```python
print("TESTING DATA VERIFICATION")
print("=========================")

print("Testing Features:", X_test_model_scaled.shape)

print("Testing Labels:", y_test.shape)

print("Total Testing Samples:", len(y_test))

print("\nClass Distribution:")

print(
    y_test.value_counts()
    .sort_index()
    .rename(index={
        0: "Malignant",
        1: "Benign"
    })
)

assert X_test_model_scaled.shape == (114, 30)
assert len(y_test) == 114
assert np.isfinite(X_test_model_scaled).all()

print("\nTesting Data Verified Successfully!")
```

```python
test_loss, test_accuracy = model.evaluate(
    X_test_model_scaled,
    y_test.to_numpy(),
    verbose=0
)

print("NEURAL NETWORK TESTING RESULTS")
print("==============================")

print(f"Testing Loss: {test_loss:.4f}")

print(f"Testing Accuracy: {test_accuracy * 100:.2f}%")

print("\nModel Evaluation Completed!")
```

```python
predicted_probabilities = model.predict(
    X_test_model_scaled,
    verbose=0
)

y_pred = np.argmax(
    predicted_probabilities,
    axis=1
)

actual_labels = y_test.to_numpy()

print("PREDICTIONS GENERATED SUCCESSFULLY")
print("==================================")

print("Actual Labels Shape:", actual_labels.shape)

print("Predicted Labels Shape:", y_pred.shape)

print("Probability Output Shape:", predicted_probabilities.shape)

print("\nFirst 10 Predictions:")
print(y_pred[:10])
```

```python
class_mapping = {
    0: "Malignant",
    1: "Benign"
}

prediction_results = pd.DataFrame({
    "Actual Label": actual_labels,
    "Predicted Label": y_pred,
    "Actual Diagnosis": [
        class_mapping[label] for label in actual_labels
    ],
    "Predicted Diagnosis": [
        class_mapping[label] for label in y_pred
    ],
    "Malignant Probability": predicted_probabilities[:, 0],
    "Benign Probability": predicted_probabilities[:, 1]
}, index=X_test.index)

prediction_results["Correct Prediction"] = (
    prediction_results["Actual Label"]
    == prediction_results["Predicted Label"]
)

print("ACTUAL VS PREDICTED CLASSIFICATIONS")
print("===================================")

display(prediction_results.head(10))
```

```python
accuracy = accuracy_score(actual_labels, y_pred)

precision_malignant = precision_score(
    actual_labels,
    y_pred,
    pos_label=0,
    zero_division=0
)

recall_malignant = recall_score(
    actual_labels,
    y_pred,
    pos_label=0,
    zero_division=0
)

f1_malignant = f1_score(
    actual_labels,
    y_pred,
    pos_label=0,
    zero_division=0
)

precision_benign = precision_score(
    actual_labels,
    y_pred,
    pos_label=1,
    zero_division=0
)

recall_benign = recall_score(
    actual_labels,
    y_pred,
    pos_label=1,
    zero_division=0
)

f1_benign = f1_score(
    actual_labels,
    y_pred,
    pos_label=1,
    zero_division=0
)

print("MODEL PERFORMANCE METRICS")
print("=========================")

print(f"Accuracy: {accuracy * 100:.2f}%")

print("\nMALIGNANT CLASS (Positive Class)")
print(f"Precision: {precision_malignant * 100:.2f}%")
print(f"Recall:    {recall_malignant * 100:.2f}%")
print(f"F1 Score:  {f1_malignant * 100:.2f}%")

print("\nBENIGN CLASS")
print(f"Precision: {precision_benign * 100:.2f}%")
print(f"Recall:    {recall_benign * 100:.2f}%")
print(f"F1 Score:  {f1_benign * 100:.2f}%")
```

```python
print("CLASSIFICATION REPORT")
print("=====================")

report = classification_report(
    actual_labels,
    y_pred,
    labels=[0, 1],
    target_names=["Malignant", "Benign"],
    digits=4,
    zero_division=0
)

print(report)
```

```python

cm = confusion_matrix(
    actual_labels,
    y_pred,
    labels=[0, 1]
)

print("CONFUSION MATRIX")
print("================")

print(cm)
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Malignant", "Benign"],
    yticklabels=["Malignant", "Benign"],
    cbar=True
)

plt.title("Confusion Matrix - Cancer Classification")

plt.xlabel("Predicted Diagnosis")
plt.ylabel("Actual Diagnosis")

plt.tight_layout()
plt.show()
```

```python
TP = cm[0, 0]
FN = cm[0, 1]
FP = cm[1, 0]
TN = cm[1, 1]

print("CONFUSION MATRIX VALUES")
print("=======================")

print("True Positives (TP):", TP)
print("False Negatives (FN):", FN)
print("False Positives (FP):", FP)
print("True Negatives (TN):", TN)

sensitivity = TP / (TP + FN) if (TP + FN) > 0 else 0

specificity = TN / (TN + FP) if (TN + FP) > 0 else 0

print("\nADDITIONAL PERFORMANCE MEASURES")

print(f"Sensitivity: {sensitivity * 100:.2f}%")
print(f"Specificity: {specificity * 100:.2f}%")
```

```python
def predict_tumor(*values):

    if len(values) != len(feature_names):
        return (
            "Invalid Input",
            "Exactly 30 numerical values are required.",
            "Please provide all tumor measurements."
        )

    if any(value is None for value in values):
        return (
            "Invalid Input",
            "One or more fields are empty.",
            "Please fill in all 30 measurement fields."
        )

    try:
        numerical_values = np.asarray(
            values,
            dtype=np.float64
        )

    except (ValueError, TypeError):
        return (
            "Invalid Input",
            "Invalid numerical values detected.",
            "Please enter valid numerical measurements."
        )

    if not np.isfinite(numerical_values).all():
        return (
            "Invalid Input",
            "Non-finite values detected.",
            "Please enter valid finite numbers."
        )

    if np.any(numerical_values < 0):
        return (
            "Invalid Input",
            "Negative measurements detected.",
            "Please enter non-negative measurement values."
        )

    input_df = pd.DataFrame(
        [numerical_values],
        columns=feature_names
    )

    scaled_input = training_scaler.transform(input_df)

    probabilities = model.predict(
        scaled_input,
        verbose=0
    )[0]

    predicted_class = int(np.argmax(probabilities))

    class_mapping = {
        0: "Malignant",
        1: "Benign"
    }

    predicted_diagnosis = class_mapping[predicted_class]

    predicted_probability = (
        float(probabilities[predicted_class]) * 100
    )

    prediction_text = (
        f"Predicted Classification: {predicted_diagnosis}\n"
        f"Predicted Class: {predicted_class}\n"
        f"Model Probability: {predicted_probability:.2f}%"
    )

    probability_text = (
        f"Malignant: {probabilities[0] * 100:.2f}%\n"
        f"Benign: {probabilities[1] * 100:.2f}%"
    )

    lower_limits = X_fit.min()
    upper_limits = X_fit.max()

    outside_range = (
        (input_df.iloc[0] < lower_limits)
        | (input_df.iloc[0] > upper_limits)
    )

    if outside_range.any():
        warning_text = (
            "Warning: Some measurements are outside "
            "the observed training-data ranges. "
            "The prediction may be unreliable."
        )
    else:
        warning_text = (
            "This is an educational AI prediction only. "
            "It is not a medical diagnosis or a substitute "
            "for professional medical evaluation."
        )

    return (
        prediction_text,
        probability_text,
        warning_text
    )


print("Cancer Prediction Function Created Successfully!")
```
